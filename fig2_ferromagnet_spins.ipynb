```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 4.5))
np.random.seed(42)

# Part A: Thermal disordered state (High Temperature)
for x in range(5):
    for y in range(5):
        angle = np.random.uniform(0, 2 * np.pi)
        ax1.quiver(x, y, float(np.cos(angle)), float(np.sin(angle)), color='black', 
                   scale=6, scale_units='inches', headwidth=4, headlength=5)
ax1.set_title(r"High Temperature ($T > T_C$)", fontsize=12, fontweight='bold', pad=15)
ax1.axis('off')

# Part B: Spontaneous symmetry breaking and aligned configuration (Low Temperature)
for x in range(5):
    for y in range(5):
        aligned_angle = 0.15
        ax2.quiver(x, y, float(np.cos(aligned_angle)), float(np.sin(aligned_angle)), color='black', 
                   scale=6, scale_units='inches', headwidth=4, headlength=5)
ax2.set_title(r"Low Temperature ($T < T_C$)", fontsize=12, fontweight='bold', pad=15)
ax2.axis('off')

plt.tight_layout()
plt.savefig("fig2_ferromagnet_spins.png", dpi=300, bbox_inches='tight')
plt.close()
print("[INFO] Figure 2 generated successfully.")
```
