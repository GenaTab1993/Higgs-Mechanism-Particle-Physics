```python
import matplotlib.pyplot as plt
import numpy as np

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 4.5))
r = np.linspace(-2.5, 2.0, 300)

# 1. Standard stable potential (Single minimum at the origin)
v_unbroken = 1.5 * r**2 + 0.2 * r**4
ax1.plot(r, v_unbroken, color='black', linewidth=1.5)
ax1.plot(0.0, 0.0, 'ko', markersize=6)  # Single stable vacuum expectation value
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['left'].set_position('center')
ax1.spines['bottom'].set_position(('data', 0))
ax1.set_title(r"Standard Potential ($\mu^2 > 0$)", fontsize=12, fontweight='bold', pad=15)
ax1.set_xticks([])
ax1.set_yticks([])
ax1.set_xlabel(r'$\phi$', loc='right', labelpad=5)
ax1.set_ylabel(r'$V(\phi)$', rotation=0, labelpad=15)

# 2. Mexican hat potential (Spontaneous symmetry breaking)
v_broken = -2.5 * r**2 + 0.6 * r**4
ax2.plot(r, v_broken, color='black', linewidth=1.5)
ax2.plot(0.0, 0.0, 'ko', markersize=6, fillstyle='none')  # Unstable local maximum
ax2.plot([-1.44, 1.44], [-2.6, -2.6], 'ro', markersize=6)  # Degenerate vacuum ground states
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.spines['left'].set_position('center')
ax2.spines['bottom'].set_position(('data', 0))
ax2.set_title(r"Mexican Hat Potential ($\mu^2 < 0$)", fontsize=12, fontweight='bold', pad=15)
ax2.set_xticks([])
ax2.set_yticks([])
ax2.set_xlabel(r'$\phi$', loc='right', labelpad=5)
ax2.set_ylabel(r'$V(\phi)$', rotation=0, labelpad=15)

plt.tight_layout()
plt.savefig("fig3_higgs_potentials.png", dpi=300, bbox_inches='tight')
plt.close()
print("[INFO] Figure 3 generated successfully.")
```
